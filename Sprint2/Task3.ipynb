**Описание задачи:**

Создать базовый класс Project в C#, который будет представлять проекты в рамках организации. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

**Требования к базовому классу Project:**

* **Атрибуты:** ID проекта (ProjectId), Название проекта (ProjectName), Статус проекта (Status).
* **Методы:**
  * `UpdateStatus()`: метод для обновления статуса проекта.
  * `GetProjectDetails()`: метод для получения деталей проекта.
  * `AssignTeamMember()`: метод для назначения командного члена проекту.

**Требования к производным классам:**

1. **ПродуктовыйПроект (ProductProject):** Должен содержать дополнительные атрибуты, такие как Ожидаемая дата завершения (ExpectedCompletionDate). Метод `UpdateStatus()` должен быть переопределен для включения ожидаемой даты завершения в статус проекта.
2. **ИсследовательскийПроект (ResearchProject):** Должен содержать дополнительные атрибуты, такие как Финансирование (FundingAmount). Метод `AssignTeamMember()` должен быть переопределен для указания специфических требований к навыкам командного члена для исследовательского проекта.
3. **ИнфраструктурныйПроект (InfrastructureProject)** (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Срок выполнения (ExecutionPeriod). Метод `GetProjectDetails()` должен быть переопределен для отображения срока выполнения проекта вместе с другими деталями проекта.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
public interface IReportable
{
    string GenerateSummaryReport();
    void PrintReport();
}

public interface IBudgetAuditable
{
    decimal Budget { get; }
    decimal CurrentExpenses { get; }
    void LogExpense(decimal amount);
    decimal GetRemainingBudget();
}

public class Project : IReportable, IBudgetAuditable
{
    private int _projectId;
    private string _projectName;
    private string _status;
    private decimal _budget;
    private decimal _currentExpenses;
    private string _priority;

    public int ProjectId
    {
        get => _projectId;
        protected set
        {
            if (value <= 0)
                throw new ArgumentException("ID проекта должен быть положительным числом!");
            _projectId = value;
        }
    }

    public string ProjectName
    {
        get => _projectName;
        protected set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Название проекта не может быть пустым!");
            _projectName = value;
        }
    }

    public string Status
    {
        get => _status;
        protected set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Статус не может быть пустым!");
            _status = value;
        }
    }

    public decimal Budget
    {
        get => _budget;
        protected set
        {
            if (value < 0)
                throw new ArgumentException("Бюджет не может быть отрицательным!");
            _budget = value;
        }
    }

    public decimal CurrentExpenses
    {
        get => _currentExpenses;
        protected set
        {
            if (value < 0)
                throw new ArgumentException("Расходы не могут быть отрицательными!");
            _currentExpenses = value;
        }
    }

    public string Priority
    {
        get => _priority;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Приоритет не может быть пустым!");
            _priority = value;
        }
    }

    public DateTime StartDate { get; protected set; }

    public List<string> TeamMembers { get; protected set; }

    public Project(int projectId, string projectName, decimal budget, string priority = "Средний", string status = "Создан")
    {
        ProjectId = projectId;
        ProjectName = projectName;
        Budget = budget;
        CurrentExpenses = 0;
        Priority = priority;
        Status = status;
        StartDate = DateTime.Now;
        TeamMembers = new List<string>();
    }

    public virtual void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"[Проект #{ProjectId} '{ProjectName}'] Статус изменен на: {newStatus}");
    }

    public virtual string GetProjectDetails()
    {
        string members = TeamMembers.Count > 0 ? string.Join(", ", TeamMembers) : "нет участников";
        return $"ID: {ProjectId} | '{ProjectName}' | Статус: {Status} | Приоритет: {Priority} | Бюджет: {Budget:N0} руб. (Остаток: {GetRemainingBudget():N0} руб.) | Команда: {members}";
    }

    public virtual void AssignTeamMember(string member)
    {
        if (!string.IsNullOrWhiteSpace(member))
        {
            TeamMembers.Add(member);
            Console.WriteLine($"[Проект #{ProjectId}] Назначен участник: {member}");
        }
    }

    public void ShareTeamMember(string memberName, Project targetProject)
    {
        if (TeamMembers.Contains(memberName))
        {
            Console.WriteLine($"--> ВЗАИМОДЕЙСТВИЕ: Проект '{ProjectName}' передает специалиста {memberName} в '{targetProject.ProjectName}'");
            targetProject.AssignTeamMember(memberName);
        }
        else
        {
            Console.WriteLine($"Специалист {memberName} не найден в проекте '{ProjectName}'");
        }
    }

    public void LogExpense(decimal amount)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма расхода должна быть больше 0!");
        CurrentExpenses += amount;
        Console.WriteLine($"[Финансы '{ProjectName}'] Зафиксирован расход: {amount:N0} руб. Всего израсходовано: {CurrentExpenses:N0} руб.");
    }

    public decimal GetRemainingBudget() => Budget - CurrentExpenses;

    public virtual string GenerateSummaryReport()
    {
        return $"[ОТЧЕТ '{ProjectName}'] ID: {ProjectId} | Дата старта: {StartDate:dd.MM.yyyy} | Остаток бюджета: {GetRemainingBudget():N0} руб. | Участников: {TeamMembers.Count}";
    }

    public void PrintReport()
    {
        Console.WriteLine(GenerateSummaryReport());
    }
}

public class ProductProject : Project
{
    private DateTime _expectedCompletionDate;
    public string TargetMarket { get; set; }
    public string ReleaseVersion { get; set; }
    public List<string> FeaturesList { get; protected set; }

    public DateTime ExpectedCompletionDate
    {
        get => _expectedCompletionDate;
        protected set
        {
            if (value < DateTime.Now.Date.AddDays(-1))
                throw new ArgumentException("Ожидаемая дата завершения не может быть в прошлом!");
            _expectedCompletionDate = value;
        }
    }

    public ProductProject(int projectId, string projectName, decimal budget, DateTime expectedCompletionDate, string targetMarket = "B2C", string status = "В разработке")
        : base(projectId, projectName, budget, "Высокий", status)
    {
        ExpectedCompletionDate = expectedCompletionDate;
        TargetMarket = targetMarket;
        ReleaseVersion = "v1.0.0";
        FeaturesList = new List<string>();
    }

    public override void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"[Продуктовый проект #{ProjectId}] Статус: {Status}. Дедлайн: {ExpectedCompletionDate:dd.MM.yyyy}, Релиз: {ReleaseVersion}");
    }

    public void AddFeature(string featureName)
    {
        FeaturesList.Add(featureName);
        Console.WriteLine($"[Продукт '{ProjectName}'] Добавлена новая фича: {featureName}");
    }

    public void PostponeDeadline(int days)
    {
        ExpectedCompletionDate = ExpectedCompletionDate.AddDays(days);
        Console.WriteLine($"[Продукт '{ProjectName}'] Дедлайн перенесен на {days} дн. Новая дата: {ExpectedCompletionDate:dd.MM.yyyy}");
    }

    public void RequestInfrastructure(InfrastructureProject infraProject)
    {
        Console.WriteLine($"--> ВЗАИМОДЕЙСТВИЕ: Продукт '{ProjectName}' запрашивает инфраструктуру у '{infraProject.ProjectName}'...");
        infraProject.AllocateResourcesForProduct(this);
    }
}

public class MobileAppProject : ProductProject
{
    public string TargetPlatform { get; set; }
    public double AppRating { get; private set; }
    public int ActiveDownloads { get; private set; }

    public MobileAppProject(int projectId, string projectName, decimal budget, DateTime expectedCompletionDate, string targetPlatform)
        : base(projectId, projectName, budget, expectedCompletionDate, "Мобильный рынок")
    {
        TargetPlatform = targetPlatform;
        AppRating = 0.0;
        ActiveDownloads = 0;
    }

    public void PublishToStore(string storeName)
    {
        UpdateStatus($"Опубликовано в {storeName}");
        Console.WriteLine($"[Мобильное приложение '{ProjectName}'] Успешно выложено в {storeName} для платформы {TargetPlatform}!");
    }

    public void RegisterDownloads(int count)
    {
        ActiveDownloads += count;
        Console.WriteLine($"[Мобильное приложение '{ProjectName}'] Новых скачиваний: +{count}. Всего: {ActiveDownloads}");
    }

    public void UpdateRating(double newRating)
    {
        AppRating = Math.Clamp(newRating, 1.0, 5.0);
        Console.WriteLine($"[Мобильное приложение '{ProjectName}'] Новый рейтинг приложения: {AppRating:F1} / 5.0");
    }
}

public class ResearchProject : Project
{
    private decimal _fundingAmount;
    public string ScientificField { get; set; }
    public int PublishedPapersCount { get; private set; }
    public int PatentsRegistered { get; private set; }

    public decimal FundingAmount
    {
        get => _fundingAmount;
        protected set
        {
            if (value < 0)
                throw new ArgumentException("Сумма финансирования не может быть отрицательной!");
            _fundingAmount = value;
        }
    }

    public ResearchProject(int projectId, string projectName, decimal budget, decimal fundingAmount, string scientificField, string status = "В исследовании")
        : base(projectId, projectName, budget, "Средний", status)
    {
        FundingAmount = fundingAmount;
        ScientificField = scientificField;
        PublishedPapersCount = 0;
        PatentsRegistered = 0;
    }

    public override void AssignTeamMember(string memberName)
    {
        string member = $"{memberName} (Исследователь в области '{ScientificField}')";
        TeamMembers.Add(member);
        Console.WriteLine($"[Исследовательский проект #{ProjectId}] Добавлен ученый: {member}");
    }

    public void PublishScientificPaper(string title)
    {
        PublishedPapersCount++;
        Console.WriteLine($"[Исследования '{ProjectName}'] Опубликована статья: '{title}'! Всего публикаций: {PublishedPapersCount}");
    }

    public void RegisterPatent(string patentNumber)
    {
        PatentsRegistered++;
        Console.WriteLine($"[Исследования '{ProjectName}'] Зарегистрирован патент #{patentNumber}! Всего патентов: {PatentsRegistered}");
    }

    public void GrantFundingTo(ResearchProject anotherResearch, decimal amount)
    {
        if (FundingAmount >= amount)
        {
            FundingAmount -= amount;
            anotherResearch.FundingAmount += amount;
            Console.WriteLine($"--> ВЗАИМОДЕЙСТВИЕ: '{ProjectName}' перевел грант {amount:N0} руб. проекту '{anotherResearch.ProjectName}'");
        }
    }
}

public class InfrastructureProject : Project
{
    private int _executionPeriodMonths;
    public int ServerCapacity { get; protected set; }
    public string DataCenterLocation { get; set; }
    public double UptimeSla { get; set; }

    public int ExecutionPeriodMonths
    {
        get => _executionPeriodMonths;
        protected set
        {
            if (value <= 0)
                throw new ArgumentException("Срок выполнения должен быть больше 0 месяцев!");
            _executionPeriodMonths = value;
        }
    }

    public InfrastructureProject(int projectId, string projectName, decimal budget, int executionPeriodMonths, int initialCapacity, string location)
        : base(projectId, projectName, budget, "Высокий", "Развертывание")
    {
        ExecutionPeriodMonths = executionPeriodMonths;
        ServerCapacity = initialCapacity;
        DataCenterLocation = location;
        UptimeSla = 99.9;
    }

    public override string GetProjectDetails()
    {
        string baseDetails = base.GetProjectDetails();
        return $"{baseDetails} | Срок: {ExecutionPeriodMonths} мес. | Серверов: {ServerCapacity} шт. | Локация: {DataCenterLocation}";
    }

    public void AllocateResourcesForProduct(ProductProject productProject)
    {
        Console.WriteLine($"[Инфраструктура '{ProjectName}'] Выделено 50 серверов для продукта '{productProject.ProjectName}'!");
        productProject.UpdateStatus("Инфраструктура подключена");
    }

    public void UpgradeCapacity(int additionalServers)
    {
        ServerCapacity += additionalServers;
        Console.WriteLine($"[Инфраструктура '{ProjectName}'] Мощности увеличены на {additionalServers} серверов. Текущая емкость: {ServerCapacity}");
    }
}

public class CloudInfrastructureProject : InfrastructureProject
{
    public string CloudProvider { get; set; }
    public bool AutoScalingEnabled { get; set; }

    public CloudInfrastructureProject(int projectId, string projectName, decimal budget, int executionMonths, int initialServers, string provider)
        : base(projectId, projectName, budget, executionMonths, initialServers, $"Облако {provider}")
    {
        CloudProvider = provider;
        AutoScalingEnabled = true;
    }

    public void EnableAutoScaling()
    {
        AutoScalingEnabled = true;
        Console.WriteLine($"[Облако '{ProjectName}'] Автомасштабирование в {CloudProvider} включено.");
    }

    public void ScaleNodes(int extraNodes)
    {
        UpgradeCapacity(extraNodes);
        Console.WriteLine($"[Облако '{ProjectName}'] Облачный кластер масштабирован в дата-центре {CloudProvider}.");
    }
}

Console.WriteLine("=== 1. СОЗДАНИЕ И ИНИЦИАЛИЗАЦИЯ ПРОЕКТОВ ===");
ProductProject webPortal = new ProductProject(1, "Корпоративный Портал", 1500000, DateTime.Now.AddMonths(4));
MobileAppProject mobileApp = new MobileAppProject(2, "Мобильный Банкинг", 2500000, DateTime.Now.AddMonths(6), "iOS & Android");
ResearchProject aiLab = new ResearchProject(3, "Лаборатория Нейросетей", 5000000, 2000000, "Машинное обучение");
CloudInfrastructureProject cloudInfra = new CloudInfrastructureProject(4, "Облачная Платформа", 3000000, 12, 100, "Yandex Cloud");

Console.WriteLine();
Console.WriteLine("=== 2. ДЕМОНСТРАЦИЯ МЕТОДОВ И ИЕРАРХИИ (СЛОЖНОЕ НАСЛЕДОВАНИЕ) ===");

mobileApp.AssignTeamMember("Дмитрий");
mobileApp.AddFeature("Авторизация по FaceID");
mobileApp.PublishToStore("App Store & Google Play");
mobileApp.RegisterDownloads(15000);
mobileApp.UpdateRating(4.8);

Console.WriteLine();

cloudInfra.AssignTeamMember("Максим");
cloudInfra.ScaleNodes(50);
cloudInfra.EnableAutoScaling();

Console.WriteLine();

aiLab.AssignTeamMember("Екатерина");
aiLab.PublishScientificPaper("Архитектура легковесных трансформеров");
aiLab.RegisterPatent("RU-2026-991A");

Console.WriteLine();
Console.WriteLine("=== 3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ЧЕРЕЗ ИНТЕРФЕЙСЫ (IReportable, IBudgetAuditable) ===");
List<IBudgetAuditable> budgetList = new List<IBudgetAuditable> { webPortal, mobileApp, aiLab, cloudInfra };
foreach (var auditable in budgetList)
{
    auditable.LogExpense(150000);
    Console.WriteLine($"Остаток средств: {auditable.GetRemainingBudget():N0} руб.");
}

Console.WriteLine();
Console.WriteLine("=== 4. ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ МЕЖДУ СОБОЙ ===");
mobileApp.RequestInfrastructure(cloudInfra);
aiLab.ShareTeamMember("Екатерина", mobileApp);

Console.WriteLine();
Console.WriteLine("=== 5. ИТОГОВЫЕ ДЕТАЛИ И ОТЧЕТЫ ПО ПРОЕКТАМ ===");
Console.WriteLine(webPortal.GetProjectDetails());
Console.WriteLine(mobileApp.GetProjectDetails());
Console.WriteLine(aiLab.GetProjectDetails());
Console.WriteLine(cloudInfra.GetProjectDetails());
```
