**Описание задачи:**

Создать базовый класс Project в C#, который будет представлять проекты в рамках организации. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

**Требования к базовому классу Project:**

* **Атрибуты:** ID проекта (ProjectId), Название проекта (ProjectName), Статус проекта (Status).
* **Методы:**
  * `UpdateStatus()`: метод для обновления статуса проекта.
  * `GetProjectDetails()`: метод для получения деталей проекта.
  * `AssignTeamMember()`: метод для назначения командного члена проекту.

**Требования к производным классам:**

1. **ПродуктовыйПроект (ProductProject):** Должен содержать дополнительные атрибуты, такие как Ожидаемая дата завершения (ExpectedCompletionDate). Метод `UpdateStatus()` должен быть переопределен для включения ожидаемой даты завершения в статус проекта.
2. **ИсследовательскийПроект (ResearchProject):** Должен содержать дополнительные атрибуты, такие как Финансирование (FundingAmount). Метод `AssignTeamMember()` должен быть переопределен для указания специфических требований к навыкам командного члена для исследовательского проекта.
3. **ИнфраструктурныйПроект (InfrastructureProject)** (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Срок выполнения (ExecutionPeriod). Метод `GetProjectDetails()` должен быть переопределен для отображения срока выполнения проекта вместе с другими деталями проекта.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

```
public class Project
{
    private int _projectId;
    private string _projectName;
    private string _status;

    public int ProjectId
    {
        get { return _projectId; }
        protected set
        {
            if (value <= 0)
                throw new ArgumentException("ID проекта должен быть больше нуля!");
            _projectId = value;
        }
    }

    public string ProjectName
    {
        get { return _projectName; }
        protected set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Название проекта не может быть пустым!");
            _projectName = value;
        }
    }

    public string Status
    {
        get { return _status; }
        protected set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Статус не может быть пустым!");
            _status = value;
        }
    }

    public List<string> TeamMembers { get; protected set; }

    public Project(int projectId, string projectName, string status = "Создан")
    {
        ProjectId = projectId;
        ProjectName = projectName;
        Status = status;
        TeamMembers = new List<string>();
    }

    public virtual void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"[Проект #{ProjectName}] Статус изменен на {newStatus}");
    }

    public virtual string GetProjectDetails()
    {
        string members = TeamMembers.Count > 0 ? string.Join(", ", TeamMembers) : "нет участников";
        return $"ID: {ProjectId} | Название: '{ProjectName}' | Статус: {Status} | Команда: {members}";
    }

    public virtual void AssignTeamMember(string member)
    {
        if (!string.IsNullOrWhiteSpace(member))
        {
            TeamMembers.Add(member);
            Console.WriteLine($"[Проект #{ProjectName}] Назначен участник: {member}");
        }
    }

    public void ShareTeamMember(string memberName, Project targetProject)
    {
        if (TeamMembers.Contains(memberName))
        {
            Console.WriteLine($"\n--> ВЗАИМОДЕЙСТВИЕ: Проект '{ProjectName}' передает специалиста {memberName} в проект '{targetProject.ProjectName}'");
            targetProject.AssignTeamMember(memberName);
        }
        else
        {
            Console.WriteLine($"Специалист {memberName} не найден в проекте '{ProjectName}'");
        }
    }
}

public class ProductProject : Project
{
    private DateTime _expectedCompletionDate;

    public DateTime ExpectedCompletionDate
    {
        get { return _expectedCompletionDate; }
        protected set
        {
            if (value < DateTime.Now.Date)
                throw new ArgumentException("Ожидаемая дата завершения не может быть в прошлом!");
            _expectedCompletionDate = value;
        }
    }

    public ProductProject(int projectId, string projectName, DateTime expectedCompletionDate, string status = "Создан") 
        : base(projectId, projectName, status)
    {
        ExpectedCompletionDate = expectedCompletionDate;
    }

    public override void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"[Продуктовый проект #{ProjectId}] Новый статус: {Status}. " +
                          $"Ожидаемая дата сдачи: {ExpectedCompletionDate:dd.MM.yyyy}");
    }

    public void RequestInfrastructure(InfrastructureProject infraProject)
    {
        Console.WriteLine($"\n--> ВЗАИМОДЕЙСТВИЕ: Продуктовый проект '{ProjectName}' запрашивает ресурсы у '{infraProject.ProjectName}'...");
        infraProject.AllocateResourcesForProduct(this);
    }
}

public class ResearchProject : Project
{
    private decimal _fundingAmount;

    public decimal FundingAmount
    {
        get { return _fundingAmount; }
        protected set
        {
            if (value < 0)
                throw new ArgumentException("Сумма финансирования не может быть отрицательной!");
            _fundingAmount = value;
        }
    }

    public ResearchProject(int projectId, string projectName, decimal fundingAmount, string status = "В исследовании") 
        : base(projectId, projectName, status)
    {
        FundingAmount = fundingAmount;
    }

    public override void AssignTeamMember(string memberName)
    {
        string memberWithSkill = $"{memberName} (Требуемая квалификация: Исследователь/Научный сотрудник)";
        TeamMembers.Add(memberWithSkill);
        Console.WriteLine($"[Исследовательский проект #{ProjectId}] Добавлен специалист: {memberWithSkill}");
    }

    public void AssignTeamMember(string memberName, string requiredSkill)
    {
        string memberWithSkill = $"{memberName} (Специализация: {requiredSkill})";
        TeamMembers.Add(memberWithSkill);
        Console.WriteLine($"[Исследовательский проект #{ProjectId}] Добавлен узкий специалист: {memberWithSkill}");
    }

    public void GrantFundingTo(ResearchProject anotherResearch, decimal amount)
    {
        if (FundingAmount >= amount)
        {
            FundingAmount -= amount;
            anotherResearch.FundingAmount += amount;
            Console.WriteLine($"\n--> ВЗАИМОДЕЙСТВИЕ: Проект '{ProjectName}' перевел {amount} руб. в бюджет проекта '{anotherResearch.ProjectName}'");
        }
    }
}

public class InfrastructureProject : Project
{
    private int _executionPeriodMonths;

    public int ExecutionPeriodMonths
    {
        get { return _executionPeriodMonths; }
        protected set
        {
            if (value <= 0)
                throw new ArgumentException("Срок выполнения должен быть больше 0 месяцев!");
            _executionPeriodMonths = value;
        }
    }

    public InfrastructureProject(int projectId, string projectName, int executionPeriodMonths, string status = "Планирование") 
        : base(projectId, projectName, status)
    {
        ExecutionPeriodMonths = executionPeriodMonths;
    }

    public override string GetProjectDetails()
    {
        string baseDetails = base.GetProjectDetails();
        return $"{baseDetails} | Срок выполнения: {ExecutionPeriodMonths} мес.";
    }

    public void AllocateResourcesForProduct(ProductProject productProject)
    {
        Console.WriteLine($"[Инфраструктурный проект '{ProjectName}'] Выделил мощности для релиза '{productProject.ProjectName}'!");
        productProject.UpdateStatus("Инфраструктура готова к релизу");
    }
}

ProductProject mobileApp = new ProductProject(1, "Мобильное Приложение", DateTime.Now.AddMonths(3));
ResearchProject aiModel = new ResearchProject(2, "ИИ Модель", 500000);
ResearchProject quantumTech = new ResearchProject(3, "Квантовые вычисления", 100000);
InfrastructureProject servers = new InfrastructureProject(4, "Дата-Центр", 6);

mobileApp.AssignTeamMember("Алексей");
aiModel.AssignTeamMember("Елена", "ML Engineer");
servers.AssignTeamMember("Сергей");

Console.WriteLine("\nВзаимодествие объектов между собой");

mobileApp.ShareTeamMember("Алексей", servers);

aiModel.GrantFundingTo(quantumTech, 150000);

mobileApp.RequestInfrastructure(servers);

Console.WriteLine("\nИтоговые детали проектов");
Console.WriteLine(mobileApp.GetProjectDetails());
Console.WriteLine(aiModel.GetProjectDetails());
Console.WriteLine(quantumTech.GetProjectDetails());
Console.WriteLine(servers.GetProjectDetails());
```
