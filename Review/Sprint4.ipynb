```polyglot-notebook
#nullable enable

public interface ISupplier
{
    string SupplierType { get; set; }
    string GetName();
    decimal GetCost();
}

public interface IEmployee
{
    string Position { get; set; }
    string GetFullName();
    decimal GetSalary();
}

public interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List<ISupplier> GetSuppliers();
    List<IEmployee> GetEmployees();
    decimal CalculateTotalExpenses();
}

public class Supplier : ISupplier
{
    public string Name { get; set; }
    public decimal Cost { get; set; }
    public string SupplierType { get; set; }

    public Supplier(string name, decimal cost, string supplierType)
    {
        Name = name;
        Cost = cost;
        SupplierType = supplierType;
    }

    public string GetName() => Name;

    public decimal GetCost() => Cost;

    public override string ToString() =>
        $"{Name} (Тип: {SupplierType}) – Стоимость: {Cost}";
}

public class Employee : IEmployee
{
    public string FullName { get; set; }
    public string Position { get; set; }
    public decimal Salary { get; set; }

    public Employee(string fullName, string position, decimal salary)
    {
        FullName = fullName;
        Position = position;
        Salary = salary;
    }

    public string GetFullName() => FullName;

    public decimal GetSalary() => Salary;

    public override string ToString() =>
        $"{FullName} – Должность: {Position} – Зарплата: {Salary}";
}

public class Enterprise : IEnterprise
{
    public string Name { get; set; }

    private readonly List<ISupplier> _suppliers = new();
    private readonly List<IEmployee> _employees = new();

    public Enterprise(string name = "Главное предприятие")
    {
        Name = name;
    }

    public void AddSupplier(ISupplier supplier)
    {
        if (supplier == null) throw new ArgumentNullException(nameof(supplier));
        _suppliers.Add(supplier);
    }

    public void AddEmployee(IEmployee employee)
    {
        if (employee == null) throw new ArgumentNullException(nameof(employee));
        _employees.Add(employee);
    }

    public List<ISupplier> GetSuppliers() => new List<ISupplier>(_suppliers);

    public List<IEmployee> GetEmployees() => new List<IEmployee>(_employees);

    public decimal CalculateTotalExpenses()
    {
        decimal supplierExpenses = _suppliers.Sum(s => s.GetCost());
        decimal employeeExpenses = _employees.Sum(e => e.GetSalary());
        return supplierExpenses + employeeExpenses;
    }
}

IEnterprise enterprise = new Enterprise("ТехноХолдинг");

enterprise.AddSupplier(new Supplier("Логистический центр", 100000m, "Логистика"));
enterprise.AddSupplier(new Supplier("Поставщик офисной мебели", 50000m, "Офисное оборудование"));

enterprise.AddEmployee(new Employee("Иван Иванов", "Менеджер", 70000m));
enterprise.AddEmployee(new Employee("Анна Петрова", "Разработчик", 95000m));

Console.WriteLine("Список поставщиков:");
int supplierIndex = 1;
foreach (var supplier in enterprise.GetSuppliers())
{
    Console.WriteLine($"{supplierIndex++}. {supplier.GetName()} – Стоимость услуг: {supplier.GetCost()}");
}

Console.WriteLine("\nСписок сотрудников:");
int employeeIndex = 1;
foreach (var employee in enterprise.GetEmployees())
{
    Console.WriteLine($"{employeeIndex++}. {employee.GetFullName()} – Должность: {employee.Position} – Зарплата: {employee.GetSalary()}");
}

Console.WriteLine($"Общие расходы: {enterprise.CalculateTotalExpenses()}");
```
