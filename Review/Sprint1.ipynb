```polyglot-notebook
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name { get => name; set => name = value; }
    public decimal Price { get => price; set => price = value; }
    public int Quantity { get => quantity; set => quantity = value; }
    public string Category { get => category; set => category = value; }
    public string Description { get => description; set => description = value; }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    public decimal GetTotalPrice() => quantity * price;

    public void UpdateQuantity(int amount) => quantity += amount;

    public string GetProductInfo()
    {
        return $"[Товар] {name} | Категория: {category} | Цена: {price} руб. | Кол-во: {quantity} шт. | Описание: {description}";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;

    public List<Product> Products { get; } = new List<Product>();

    public string Name { get => name; set => name = value; }
    public string EmployeeId { get => employeeId; set => employeeId = value; }
    public decimal Salary { get => salary; set => salary = value; }
    public string ContactInfo { get => contactInfo; set => contactInfo = value; }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        Products.Add(product);
        Console.WriteLine($"Товар \"{product.Name}\" добавлен к продавцу {name}.");
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.UpdateQuantity(-quantity);
            Console.WriteLine($"Продавец {name} продал {quantity} шт. товара \"{product.Name}\".");
        }
        else
        {
            Console.WriteLine($"Недостаточно товара \"{product.Name}\" на складе!");
        }
    }

    public string GetSellerInfo()
    {
        return $"[Продавец] {name} | ID: {employeeId} | Зарплата: {salary} руб. | Контакты: {contactInfo}";
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;

    private List<Seller> sellers = new List<Seller>();

    public string StoreName { get => storeName; set => storeName = value; }
    public string Location { get => location; set => location = value; }

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
        Console.WriteLine($"Продавец {seller.Name} нанят в магазин \"{storeName}\".");
    }

    public void ListProducts()
    {
        Console.WriteLine($"\n--- Список товаров в магазине \"{storeName}\" ---");
    
        if (sellers.Count == 0)
        {
            Console.WriteLine("В магазине нет продавцов, а значит и товаров.");
            return;
        }

        bool hasProducts = false;

        foreach (var seller in sellers)
        {
            foreach (var product in seller.Products)
            {
                Console.WriteLine(product.GetProductInfo());
                hasProducts = true;
            }
        }

        if (!hasProducts)
        {
            Console.WriteLine("На складе продавцов пока нет товаров.");
        }
    }

    public string GetStoreInfo()
    {
        return $"[Магазин] \"{storeName}\" | Адрес: {location} | Режим работы: {storeHours}";
    }
}



Product phone = new Product(
    name: "Смартфон",
    price: 50000m,
    quantity: 5,
    category: "Электроника",
    description: "Флагманский телефон 256GB"
);

Seller seller = new Seller(
    name: "Алексей Петров",
    employeeId: "EMP-102",
    salary: 60000m,
    contactInfo: "+7 (900) 123-45-67"
);

Store store = new Store(
    storeName: "ЭлектроМир",
    location: "ул. Пушкина, д. 15",
    storeHours: "10:00 - 22:00"
);

Console.WriteLine("МЕТОДЫ GET");
Console.WriteLine(store.GetStoreInfo());
Console.WriteLine(seller.GetSellerInfo());
Console.WriteLine(phone.GetProductInfo());
Console.WriteLine($"Общая стоимость товара: {phone.GetTotalPrice()} руб.");

Console.WriteLine("\nВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ");
store.AddSeller(seller);
seller.AddProduct(phone);
store.ListProducts();

Console.WriteLine("ПРОВЕРКА ПРОДАЖИ И ОБНОВЛЕНИЯ КОЛИЧЕСТВА");
seller.SellProduct(phone, 2);

Console.WriteLine(phone.GetProductInfo());
Console.WriteLine($"Новая общая стоимость товара: {phone.GetTotalPrice()} руб.");
```
