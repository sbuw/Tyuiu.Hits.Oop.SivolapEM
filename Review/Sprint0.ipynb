```polyglot-notebook
public class Plant
{
    public string Name { get; private set; }
    public string Type { get; private set; }
    public float Height { get; private set; }

    public Plant(string name, string type, float height){
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom(){
        Console.WriteLine($"Растение {Name} цветет.");
    }

    public void Grow(float growth){
        Height += growth;
        Console.WriteLine($"Растение выросло на {growth} см. Новая высота {Height} см.");
    }
}

public class Flower: Plant
{
    public string Color { get; private set; }
    public int PetalCount { get; private set; }

    public Flower(string name, float height, string color, int petalCount) : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать");
    }
}

public class Tree: Plant
{
    public int Age { get; private set; }
    public string LeafType { get; private set; }

    public Tree(string name, float height, int age, string leafType) : base(name, "Дерево", height)
    {
        LeafType = leafType;
        Age = age;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowOlder(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} увеличило возраст на {years} лет. Теперь его возраст {Age} лет.");
    }
}

Flower rose = new Flower("Роза", 25.5f, "Красный", 12);
Tree oak = new Tree("Дуб", 300.0f, 10, "Лиственные");

Console.WriteLine("=== ТЕСТИРОВАНИЕ ЦВЕТКА ===");
rose.Bloom();
rose.Grow(5.0f);
rose.Wither();

Console.WriteLine("\n=== ТЕСТИРОВАНИЕ ДЕРЕВА ===");
oak.Bloom();
oak.Grow(15.0f);
oak.GrowOlder(2);

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ===");
Plant[] garden = new Plant[] { rose, oak };

foreach (Plant plant in garden)
{
    plant.Bloom();
}
```
