```polyglot-notebook
public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"[Сотрудник] Имя: {Name} | Должность: {Position} | Зарплата: {Salary} руб.");
    }
}

public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        Console.WriteLine($"[Менеджер] Имя: {Name} | Должность: {Position} | Зарплата: {Salary} руб. | Отдел: {Department}");
    }
}

public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string[] GetTeamMembers();
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; }
    private List<string> teamMembers = new List<string>();

    public ProjectManager(string name, string position, double salary, string department, string projectName)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
    }

    public ProjectManager(string name, string position, double salary, string department, string projectName, string[] members)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        if (members != null)
        {
            teamMembers.AddRange(members);
        }
    }

    public void AddTeamMember(string member)
    {
        teamMembers.Add(member);
    }

    public override void PrintInfo()
    {
        Console.WriteLine($"[Проектный менеджер] Имя: {Name} | Должность: {Position} | Зарплата: {Salary} руб. | Отдел: {Department} | Проект: {ProjectName}");
    }

    public string GenerateReport()
    {
        return $"[Отчет по проекту] Проект: \"{ProjectName}\" | Менеджер: {Name} | Отдел: {Department} | Статус: В разработке | Количество сотрудников: {teamMembers.Count}";
    }

    public string[] GetTeamMembers()
    {
        return teamMembers.ToArray();
    }
}

Employee employee = new Employee("Иван Иванов", "Инженер-программист", 85000.0);
Manager manager = new Manager("Елена Смирнова", "Руководитель IT-отдела", 140000.0, "Департамент разработки");
ProjectManager projectManager = new ProjectManager(
    "Сергей Васильев",
    "Ведущий проектный менеджер",
    180000.0,
    "Департамент разработки",
    "Система управления предприятием (ERP)",
    new string[] { "Алексей Попов (Frontend-разработчик)", "Дмитрий Соколов (Backend-разработчик)", "Анна Кузнецова (QA-инженер)" }
);

Console.WriteLine("ИНФОРМАЦИЯ О СОТРУДНИКАХ");
employee.PrintInfo();
manager.PrintInfo();
projectManager.PrintInfo();

Console.WriteLine("\nТЕСТИРОВАНИЕ ИНТЕРФЕЙСОВ ДЛЯ PROJECTMANAGER");
Console.WriteLine("Генерация отчета (IReportable):");
Console.WriteLine(projectManager.GenerateReport());

Console.WriteLine("\nЧлены команды (ITeamLeader):");
string[] members = projectManager.GetTeamMembers();
foreach (string member in members)
{
    Console.WriteLine($"- {member}");
}

Console.WriteLine("\nВЫЗОВ ЧЕРЕЗ ИНТЕРФЕЙСНЫЕ ССЫЛКИ");
IReportable reportable = projectManager;
ITeamLeader teamLeader = projectManager;
Console.WriteLine($"IReportable.GenerateReport(): {reportable.GenerateReport()}");
Console.WriteLine($"ITeamLeader.GetTeamMembers().Length: {teamLeader.GetTeamMembers().Length}");

Console.WriteLine("\nДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА");
Employee[] company = new Employee[] { employee, manager, projectManager };
foreach (Employee emp in company)
{
    emp.PrintInfo();
}
```
