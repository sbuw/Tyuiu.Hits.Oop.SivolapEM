**Описание задачи:**

Создать базовый класс Project в C#, который будет представлять проекты в рамках организации. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

**Требования к базовому классу Project:**

* **Атрибуты:** ID проекта (ProjectId), Название проекта (ProjectName), Статус проекта (Status).
* **Методы:**
  * `UpdateStatus()`: метод для обновления статуса проекта.
  * `GetProjectDetails()`: метод для получения деталей проекта.
  * `AssignTeamMember()`: метод для назначения командного члена проекту.

**Требования к производным классам:**

1. **ПродуктовыйПроект (ProductProject):** Должен содержать дополнительные атрибуты, такие как Ожидаемая дата завершения (ExpectedCompletionDate). Метод `UpdateStatus()` должен быть переопределен для включения ожидаемой даты завершения в статус проекта.
2. **ИсследовательскийПроект (ResearchProject):** Должен содержать дополнительные атрибуты, такие как Финансирование (FundingAmount). Метод `AssignTeamMember()` должен быть переопределен для указания специфических требований к навыкам командного члена для исследовательского проекта.
3. **ИнфраструктурныйПроект (InfrastructureProject)** (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Срок выполнения (ExecutionPeriod). Метод `GetProjectDetails()` должен быть переопределен для отображения срока выполнения проекта вместе с другими деталями проекта.

```
public interface IInternalAuditable
{
    void PerformAudit();
    string GetAuditStatus();
}

public interface IArchivable
{
    void ArchiveProject();
    bool IsArchived { get; }
}

public interface IProjectNotifier
{
    void SendNotification(string recipient, string message);
}

public class Project : IInternalAuditable, IArchivable
{
    private int _projectId;
    private string _projectName;
    private string _status;
    private bool _isArchived;
    private bool _auditPassed;

    public int ProjectId
    {
        get => _projectId;
        protected set
        {
            if (value <= 0)
                throw new ArgumentException("ID проекта должен быть положительным числом!");
            _projectId = value;
        }
    }

    public string ProjectName
    {
        get => _projectName;
        protected set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Название проекта не может быть пустым!");
            _projectName = value;
        }
    }

    public string Status
    {
        get => _status;
        protected set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Статус не может быть пустым!");
            _status = value;
        }
    }

    public string ProjectManager { get; set; }
    public string RiskLevel { get; set; }
    public decimal Budget { get; protected set; }
    public List<string> TeamMembers { get; protected set; }

    public Project(int projectId, string projectName, decimal budget, string manager = "Не назначен", string status = "Создан")
    {
        ProjectId = projectId;
        ProjectName = projectName;
        Budget = budget;
        ProjectManager = manager;
        RiskLevel = "Низкий";
        Status = status;
        _isArchived = false;
        _auditPassed = false;
        TeamMembers = new List<string>();
    }

    public virtual void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"[Проект #{ProjectId} '{ProjectName}'] Статус изменен на: {newStatus}");
    }

    public virtual string GetProjectDetails()
    {
        string members = TeamMembers.Count > 0 ? string.Join(", ", TeamMembers) : "нет участников";
        return $"ID: {ProjectId} | '{ProjectName}' | Менеджер: {ProjectManager} | Риск: {RiskLevel} | Статус: {Status} | Команда: {members}";
    }

    public virtual void AssignTeamMember(string member)
    {
        if (!string.IsNullOrWhiteSpace(member))
        {
            TeamMembers.Add(member);
            Console.WriteLine($"[Проект #{ProjectId}] Назначен участник: {member}");
        }
    }

    public void SetManager(string managerName)
    {
        ProjectManager = managerName;
        Console.WriteLine($"[Проект #{ProjectId} '{ProjectName}'] Назначен руководитель: {managerName}");
    }

    public void UpdateRiskLevel(string newRiskLevel)
    {
        RiskLevel = newRiskLevel;
        Console.WriteLine($"[Проект #{ProjectId} '{ProjectName}'] Уровень риска обновлен: {newRiskLevel}");
    }

    void IInternalAuditable.PerformAudit()
    {
        _auditPassed = true;
        Console.WriteLine($"[Явная реализация IInternalAuditable] Аудит проекта '{ProjectName}' успешно пройден.");
    }

    string IInternalAuditable.GetAuditStatus()
    {
        return _auditPassed ? "Соответствует стандартам качества" : "Требует проверки";
    }

    void IArchivable.ArchiveProject()
    {
        _isArchived = true;
        UpdateStatus("Архивирован");
        Console.WriteLine($"[Явная реализация IArchivable] Проект '{ProjectName}' перемещен в долгосрочный архив.");
    }

    bool IArchivable.IsArchived => _isArchived;
}

public class ProductProject : Project
{
    public DateTime ExpectedCompletionDate { get; set; }
    public string ProductOwner { get; set; }

    public ProductProject(int projectId, string projectName, decimal budget, DateTime completionDate, string owner)
        : base(projectId, projectName, budget, owner, "В разработке")
    {
        ExpectedCompletionDate = completionDate;
        ProductOwner = owner;
    }

    public override void UpdateStatus(string newStatus)
    {
        base.UpdateStatus(newStatus);
        Console.WriteLine($"[Продукт '{ProjectName}'] Ожидаемый релиз: {ExpectedCompletionDate:dd.MM.yyyy}");
    }
}

public class ResearchProject : Project
{
    public decimal FundingAmount { get; set; }
    public string LeadResearcher { get; set; }

    public ResearchProject(int projectId, string projectName, decimal budget, decimal fundingAmount, string leadResearcher)
        : base(projectId, projectName, budget, leadResearcher, "Исследование")
    {
        FundingAmount = fundingAmount;
        LeadResearcher = leadResearcher;
    }

    public override void AssignTeamMember(string member)
    {
        string role = $"{member} (Исследовательская группа {LeadResearcher})";
        TeamMembers.Add(role);
        Console.WriteLine($"[Исследования #{ProjectId}] Назначен научный сотрудник: {role}");
    }
}

public class InfrastructureProject : Project
{
    public int ServerNodes { get; set; }

    public InfrastructureProject(int projectId, string projectName, decimal budget, int serverNodes)
        : base(projectId, projectName, budget, "DevOps Team", "Развертывание")
    {
        ServerNodes = serverNodes;
    }
}

public class ConsoleProjectNotifier : IProjectNotifier
{
    public void SendNotification(string recipient, string message)
    {
        Console.WriteLine($"[УВЕДОМЛЕНИЕ для '{recipient}']: {message}");
    }
}

public class ProjectManagementService
{
    private readonly Project _project;
    private readonly IProjectNotifier _notifier;

    public ProjectManagementService(Project project, IProjectNotifier notifier)
    {
        _project = project ?? throw new ArgumentNullException(nameof(project));
        _notifier = notifier ?? throw new ArgumentNullException(nameof(notifier));
    }

    public void ProcessProjectMilestone(string milestoneName)
    {
        Console.WriteLine($"=== Выполнение контрольной точки '{milestoneName}' для '{_project.ProjectName}' ===");
        _project.UpdateStatus($"Этап '{milestoneName}' завершен");

        IInternalAuditable auditable = _project;
        auditable.PerformAudit();

        _notifier.SendNotification("PMO & Stakeholders",
            $"Проект '{_project.ProjectName}' успешно прошел веху '{milestoneName}'. Аудит: {auditable.GetAuditStatus()}");
        Console.WriteLine();
    }

    public void CompleteAndArchive()
    {
        Console.WriteLine($"=== Завершение и архивация проекта '{_project.ProjectName}' ===");
        IArchivable archivable = _project;
        archivable.ArchiveProject();

        _notifier.SendNotification("Архивный отдел",
            $"Проект ID: {_project.ProjectId} закрыт и заархивирован.");
        Console.WriteLine();
    }
}

ProductProject mobileBanking = new ProductProject(101, "Мобильный Банкинг 2.0", 5000000, DateTime.Now.AddMonths(5), "Александр");
mobileBanking.AssignTeamMember("Дмитрий");
mobileBanking.UpdateRiskLevel("Средний");

Console.WriteLine();
Console.WriteLine("=== 1. ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСОВ ===");

IInternalAuditable auditRef = mobileBanking;
auditRef.PerformAudit();
Console.WriteLine($"Статус аудита: {auditRef.GetAuditStatus()}");

IArchivable archiveRef = mobileBanking;
Console.WriteLine($"Проект заархивирован до вызова? {archiveRef.IsArchived}");

Console.WriteLine();
Console.WriteLine("=== 2. УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ (DEPENDENCY INJECTION) ===");

IProjectNotifier notifier = new ConsoleProjectNotifier();

ProjectManagementService pmService = new ProjectManagementService(mobileBanking, notifier);

pmService.ProcessProjectMilestone("Бета-тестирование релизной сборки");
pmService.CompleteAndArchive();
```
