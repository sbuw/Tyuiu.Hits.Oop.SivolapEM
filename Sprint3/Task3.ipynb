**Описание задачи:**

Создать базовый класс Project в C#, который будет представлять проекты в рамках организации. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

**Требования к базовому классу Project:**

* **Атрибуты:** ID проекта (ProjectId), Название проекта (ProjectName), Статус проекта (Status).
* **Методы:**
  * `UpdateStatus()`: метод для обновления статуса проекта.
  * `GetProjectDetails()`: метод для получения деталей проекта.
  * `AssignTeamMember()`: метод для назначения командного члена проекту.

**Требования к производным классам:**

1. **ПродуктовыйПроект (ProductProject):** Должен содержать дополнительные атрибуты, такие как Ожидаемая дата завершения (ExpectedCompletionDate). Метод `UpdateStatus()` должен быть переопределен для включения ожидаемой даты завершения в статус проекта.
2. **ИсследовательскийПроект (ResearchProject):** Должен содержать дополнительные атрибуты, такие как Финансирование (FundingAmount). Метод `AssignTeamMember()` должен быть переопределен для указания специфических требований к навыкам командного члена для исследовательского проекта.
3. **ИнфраструктурныйПроект (InfrastructureProject)** (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Срок выполнения (ExecutionPeriod). Метод `GetProjectDetails()` должен быть переопределен для отображения срока выполнения проекта вместе с другими деталями проекта.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
public interface IReportable
{
    string GenerateSummaryReport();
    void PrintReport();
}

public interface IBudgetAuditable
{
    decimal Budget { get; }
    decimal CurrentExpenses { get; }
    void LogExpense(decimal amount);
    decimal GetRemainingBudget();
}

public class Project : IReportable, IBudgetAuditable
{
    private int _projectId;
    private string _projectName;
    private string _status;
    private decimal _budget;
    private decimal _currentExpenses;
    private string _priority;

    public int ProjectId
    {
        get => _projectId;
        protected set
        {
            if (value <= 0)
                throw new ArgumentException("ID проекта должен быть положительным числом!");
            _projectId = value;
        }
    }

    public string ProjectName
    {
        get => _projectName;
        protected set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Название проекта не может быть пустым!");
            _projectName = value;
        }
    }

    public string Status
    {
        get => _status;
        protected set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Статус не может быть пустым!");
            _status = value;
        }
    }

    public decimal Budget
    {
        get => _budget;
        protected set
        {
            if (value < 0)
                throw new ArgumentException("Бюджет не может быть отрицательным!");
            _budget = value;
        }
    }

    public decimal CurrentExpenses
    {
        get => _currentExpenses;
        protected set
        {
            if (value < 0)
                throw new ArgumentException("Расходы не могут быть отрицательными!");
            _currentExpenses = value;
        }
    }

    public string Priority
    {
        get => _priority;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Приоритет не может быть пустым!");
            _priority = value;
        }
    }

    public DateTime StartDate { get; protected set; }
    public List<string> TeamMembers { get; protected set; }

    public Project(int projectId, string projectName, decimal budget, string priority = "Средний", string status = "Создан")
    {
        ProjectId = projectId;
        ProjectName = projectName;
        Budget = budget;
        CurrentExpenses = 0;
        Priority = priority;
        Status = status;
        StartDate = DateTime.Now;
        TeamMembers = new List<string>();
    }

    public virtual void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"[Проект #{ProjectId} '{ProjectName}'] Статус изменен на: {newStatus}");
    }

    public virtual string GetProjectDetails()
    {
        string members = TeamMembers.Count > 0 ? string.Join(", ", TeamMembers) : "нет участников";
        return $"ID: {ProjectId} | '{ProjectName}' | Статус: {Status} | Бюджет: {Budget:N0} руб. (Остаток: {GetRemainingBudget():N0} руб.) | Участников: {TeamMembers.Count}";
    }

    public virtual void ExecuteStage()
    {
        Console.WriteLine($"[Проект #{ProjectId} '{ProjectName}'] Выполняется базовый этап жизненного цикла проекта.");
    }

    public virtual void AssignTeamMember(string member)
    {
        if (!string.IsNullOrWhiteSpace(member))
        {
            TeamMembers.Add(member);
            Console.WriteLine($"[Проект #{ProjectId}] Добавлен участник: {member}");
        }
    }

    public void AssignTeamMember(string member, string role)
    {
        string formatted = $"{member} ({role})";
        TeamMembers.Add(formatted);
        Console.WriteLine($"[Проект #{ProjectId}] Назначен специалист с ролью: {formatted}");
    }

    public void AssignTeamMember(List<string> members)
    {
        foreach (var m in members)
        {
            AssignTeamMember(m);
        }
    }

    public void LogExpense(decimal amount)
    {
        if (amount <= 0)
            throw new ArgumentException("Сумма расхода должна быть больше 0!");
        CurrentExpenses += amount;
        Console.WriteLine($"[Финансы '{ProjectName}'] Расход: {amount:N0} руб. Остаток: {GetRemainingBudget():N0} руб.");
    }

    public void LogExpense(decimal amount, string category)
    {
        LogExpense(amount);
        Console.WriteLine($"[Финансы '{ProjectName}'] Категория расхода: '{category}'.");
    }

    public void AllocateBudget(decimal additionalAmount)
    {
        Budget += additionalAmount;
        Console.WriteLine($"[Бюджет '{ProjectName}'] Увеличен на {additionalAmount:N0} руб. Общий бюджет: {Budget:N0} руб.");
    }

    public void AllocateBudget(decimal additionalAmount, string justification)
    {
        AllocateBudget(additionalAmount);
        Console.WriteLine($"[Бюджет '{ProjectName}'] Обоснование расширения: {justification}");
    }

    public decimal GetRemainingBudget() => Budget - CurrentExpenses;

    public virtual string GenerateSummaryReport()
    {
        return $"[ОТЧЕТ '{ProjectName}'] ID: {ProjectId} | Дата: {StartDate:dd.MM.yyyy} | Бюджет: {Budget:N0} руб. | Команда: {TeamMembers.Count} чел.";
    }

    public void PrintReport()
    {
        Console.WriteLine(GenerateSummaryReport());
    }
}

public class ProductProject : Project
{
    public DateTime ExpectedCompletionDate { get; protected set; }
    public string ReleaseVersion { get; set; }
    public List<string> Features { get; protected set; }

    public ProductProject(int projectId, string projectName, decimal budget, DateTime expectedCompletionDate)
        : base(projectId, projectName, budget, "Высокий", "Разработка")
    {
        ExpectedCompletionDate = expectedCompletionDate;
        ReleaseVersion = "v1.0.0";
        Features = new List<string>();
    }

    public override void ExecuteStage()
    {
        base.ExecuteStage();
        Console.WriteLine($"[Продукт '{ProjectName}'] Проводится спринт разработки фич. Подготовка к релизу {ReleaseVersion}.");
    }

    public override void UpdateStatus(string newStatus)
    {
        base.UpdateStatus(newStatus);
        Console.WriteLine($"[Продукт '{ProjectName}'] Ожидаемый дедлайн: {ExpectedCompletionDate:dd.MM.yyyy}");
    }

    public void AddFeature(string featureName)
    {
        Features.Add(featureName);
        Console.WriteLine($"[Продукт '{ProjectName}'] Добавлена новая фича: {featureName}");
    }
}

public class ResearchProject : Project
{
    public decimal FundingAmount { get; protected set; }
    public string ResearchTopic { get; set; }

    public ResearchProject(int projectId, string projectName, decimal budget, decimal fundingAmount, string topic)
        : base(projectId, projectName, budget, "Средний", "Анализ данных")
    {
        FundingAmount = fundingAmount;
        ResearchTopic = topic;
    }

    public override void ExecuteStage()
    {
        base.ExecuteStage();
        Console.WriteLine($"[Исследование '{ProjectName}'] Проводятся научные эксперименты по теме: '{ResearchTopic}'.");
    }

    public override void AssignTeamMember(string member)
    {
        string specialized = $"{member} (Исследователь: {ResearchTopic})";
        TeamMembers.Add(specialized);
        Console.WriteLine($"[Исследования #{ProjectId}] Назначен научный сотрудник: {specialized}");
    }
}

public class InfrastructureProject : Project
{
    public int ServerCapacity { get; protected set; }

    public InfrastructureProject(int projectId, string projectName, decimal budget, int serverCapacity)
        : base(projectId, projectName, budget, "Высокий", "Инфраструктура")
    {
        ServerCapacity = serverCapacity;
    }

    public override void ExecuteStage()
    {
        base.ExecuteStage();
        Console.WriteLine($"[Инфраструктура '{ProjectName}'] Настройка и мониторинг кластера из {ServerCapacity} серверов.");
    }
}

public class ProjectPortfolio<T> where T : Project
{
    private List<T> _projects = new List<T>();
    public string PortfolioName { get; set; }

    public ProjectPortfolio(string portfolioName)
    {
        PortfolioName = portfolioName;
    }

    public void AddProject(T project)
    {
        _projects.Add(project);
        Console.WriteLine($"[Портфель '{PortfolioName}'] Зарегистрирован проект '{project.ProjectName}' (ID: {project.ProjectId})");
    }

    public bool RemoveProject(int projectId)
    {
        var item = FindById(projectId);
        if (item != null)
        {
            _projects.Remove(item);
            Console.WriteLine($"[Портфель '{PortfolioName}'] Проект '{item.ProjectName}' удален.");
            return true;
        }
        return false;
    }

    public T FindById(int projectId)
    {
        return _projects.FirstOrDefault(p => p.ProjectId == projectId);
    }

    public decimal CalculateTotalBudget()
    {
        return _projects.Sum(p => p.Budget);
    }

    public void ExecuteAllProjectStages()
    {
        Console.WriteLine($"--- Выполнение этапов проектов портфеля '{PortfolioName}' ---");
        foreach (var project in _projects)
        {
            project.ExecuteStage();
            Console.WriteLine();
        }
    }

    public void DisplayAll()
    {
        Console.WriteLine($"=== Портфель '{PortfolioName}' (Всего проектов: {_projects.Count}, Бюджет: {CalculateTotalBudget():N0} руб.) ===");
        foreach (var p in _projects)
        {
            Console.WriteLine(p.GetProjectDetails());
        }
    }
}

Console.WriteLine("=== 1. ОБОБЩЕННЫЙ ПОРТФЕЛЬ ПРОЕКТОВ (GENERIC) ===");
ProjectPortfolio<Project> companyPortfolio = new ProjectPortfolio<Project>("ХИТС Корпоративный Портфель");

ProductProject app = new ProductProject(1, "Мобильный Клиент", 2000000, DateTime.Now.AddMonths(4));
ResearchProject ai = new ResearchProject(2, "Нейросетевой Ассистент", 3500000, 1500000, "Deep Learning");
InfrastructureProject cloud = new InfrastructureProject(3, "Облачный Кластер", 1500000, 64);

companyPortfolio.AddProject(app);
companyPortfolio.AddProject(ai);
companyPortfolio.AddProject(cloud);

Console.WriteLine();
companyPortfolio.DisplayAll();

Console.WriteLine();
Console.WriteLine("=== 2. ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ МЕТОДОВ (OVERLOADING) ===");

app.AssignTeamMember("Иван");
app.AssignTeamMember("Ольга", "Lead UI/UX Designer");
app.AssignTeamMember(new List<string> { "Сергей (QA)", "Анна (DevOps)" });

Console.WriteLine();

app.LogExpense(50000);
app.LogExpense(120000, "Маркетинг и реклама");
app.AllocateBudget(300000, "Расширение штата тестировщиков");

Console.WriteLine();
Console.WriteLine("=== 3. ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА И ПЕРЕКРЫТИЯ (OVERRIDE + BASE) ===");
companyPortfolio.ExecuteAllProjectStages();

Console.WriteLine();

ProjectPortfolio<ProductProject> productHub = new ProjectPortfolio<ProductProject>("Продуктовый Хаб");
productHub.AddProject(app);
Console.WriteLine($"Суммарный бюджет продуктового хаба: {productHub.CalculateTotalBudget():N0} руб.");
```
